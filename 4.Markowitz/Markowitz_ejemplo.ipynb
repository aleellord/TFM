```python
modules = ["seaborn", "yfinance","PyPortfolioOpt"]

import sys
import os.path
from subprocess import check_call
import importlib
import os

def instala(modules):
  print("Instalando módulos")
  for m in modules:
      # para el import quitamos [...] y ==...
      p = m.find("[")
      mi = m if p==-1 else m[:p]
      p = mi.find("==")
      mi = mi if p==-1 else mi[:p]

      torch_loader = importlib.util.find_spec(mi)
      if torch_loader is not None:
          print(m," encontrado")
      else:
          print(m," No encontrado, instalando...",end="")  
          try:        
            r = check_call([sys.executable, "-m", "pip", "install", "--user", m])
            print("¡hecho!")
          except:
            print("¡Problema al instalar ",m,"! ¿seguro que el módulo existe?",sep="")
              
  print("¡Terminado!")

instala(modules) 
```

```python
import math
import numpy as np
import pandas as pd
import yfinance as yf

import warnings

warnings.filterwarnings(
    "ignore",
    message=".*Timestamp.utcnow is deprecated.*",
    category=pd.errors.Pandas4Warning,
    module="yfinance.*"
)

tickers = list(ucits.keys())
start_date = "2022-01-01"

raw = yf.download(
    tickers=tickers,
    start=start_date,
    auto_adjust=False,
    actions=False,
    group_by="ticker",
    threads=True,
    progress=False,
)

# -------- 1) EXTRAER PRECIOS, CLOSE Y VOLUMEN (robusto a MultiIndex invertido) --------
if not isinstance(raw.columns, pd.MultiIndex):
    # un solo ticker
    col_price = "Adj Close" if "Adj Close" in raw.columns else "Close"
    prices = raw[[col_price]].copy()
    prices.columns = [tickers[0]]

    close_prices = raw[["Close"]].copy() if "Close" in raw.columns else prices.copy()
    close_prices.columns = [tickers[0]]

    volume = raw[["Volume"]].copy() if "Volume" in raw.columns else pd.DataFrame(index=raw.index, columns=[tickers[0]])
else:
    lvl0 = set(raw.columns.get_level_values(0))
    lvl1 = set(raw.columns.get_level_values(1))

    if any(t in lvl0 for t in tickers):
        ticker_level, field_level = 0, 1
    elif any(t in lvl1 for t in tickers):
        ticker_level, field_level = 1, 0
    else:
        raise ValueError("No encuentro tickers en las columnas de 'raw'. Revisa raw.columns.")

    fields = set(raw.columns.get_level_values(field_level))

    def get_field(field_name: str) -> pd.DataFrame:
        return raw.xs(field_name, axis=1, level=field_level)

    # Para análisis de precios/retornos
    if "Adj Close" in fields:
        prices = get_field("Adj Close")
    else:
        prices = get_field("Close")

    # Para calcular dinero negociado
    if "Close" in fields:
        close_prices = get_field("Close")
    else:
        close_prices = prices.copy()

    volume = get_field("Volume") if "Volume" in fields else pd.DataFrame(index=raw.index)

    prices = prices.reindex(columns=[t for t in tickers if t in prices.columns])
    close_prices = close_prices.reindex(columns=prices.columns)
    volume = volume.reindex(columns=prices.columns)

prices.index.name = "Date"
prices = prices.sort_index()
close_prices = close_prices.sort_index()
volume = volume.sort_index()

# -------- 2) FILTRO NaNs: fuera columnas con >20% NaN --------
min_non_nan = math.ceil(len(prices) * 0.80)
prices = prices.dropna(axis=1, thresh=min_non_nan)
close_prices = close_prices.reindex(columns=prices.columns)
volume = volume.reindex(columns=prices.columns)

# -------- 3) FILTRO “historia desde el inicio REAL” --------
start_ts = prices.index.min()
first_valid = prices.apply(lambda s: s.first_valid_index())
prices = prices.loc[:, first_valid <= start_ts]
close_prices = close_prices.reindex(columns=prices.columns)
volume = volume.reindex(columns=prices.columns)

# -------- 4) FILTRO LIQUIDEZ: eliminar el 20% que menos dinero mueve --------
# Dinero negociado aproximado por día = Close * Volume
money_volume = close_prices * volume

# tratamos 0 como NaN para no premiar días sin negociación
money_vol_mean = money_volume.replace(0, np.nan).mean(skipna=True)

# si algún ticker no tiene dato nunca, lo mandamos al fondo
money_vol_mean = money_vol_mean.fillna(0)

cut = money_vol_mean.quantile(0.20)
keep = money_vol_mean[money_vol_mean > cut].index

prices = prices[keep]
close_prices = close_prices[keep]
volume = volume[keep]


# -------- 5) Relleno de huecos razonable (sin cargarnos todo todo) --------
prices = prices.ffill().dropna()

print("prices shape:", prices.shape)
print("volume shape:", volume.shape)
print("primera fecha:", prices.index.min(), "última:", prices.index.max())
```

```python
import numpy as np
import pandas as pd

from pypfopt import risk_models, EfficientFrontier, objective_functions
from pypfopt.expected_returns import returns_from_prices

import cvxpy as cp


def pick_solver():
    solvers = set(cp.installed_solvers())

    # Orden práctico para probar en este tipo de problema
    if "PIQP" in solvers:
        return "PIQP", {"backend": "sparse", "eps_abs": 1e-6, "eps_rel": 1e-6, "max_iter": 10000}
    elif "CLARABEL" in solvers:
        return "CLARABEL", {}
    elif "PROXQP" in solvers:
        return "PROXQP", {"backend": "sparse", "eps_abs": 1e-6, "eps_rel": 1e-6, "max_iter": 5000}
    elif "OSQP" in solvers:
        return "OSQP", {"max_iter": 20000, "eps_abs": 1e-4, "eps_rel": 1e-4}
    elif "SCS" in solvers:
        return "SCS", {"max_iters": 20000, "eps": 1e-4}
    else:
        raise RuntimeError(f"Sin solver útil. Instalados: {sorted(solvers)}")


def markowitz_from_prices(
    prices: pd.DataFrame,
    lam: float = 10.0,              # prueba 5, 10, 20, 50...
    max_weight: float = 0.25,       # 25% por activo (evita concentración)
    l2_gamma: float = 0.0,          # 0.05–0.5 suele ayudar a diversificar
    topk: int = 10
):
    # --- Limpieza mínima (asumo que 'prices' ya viene bastante limpio) ---
    prices = prices.copy()
    prices = prices.replace([np.inf, -np.inf], np.nan).ffill()
    prices = prices.dropna(axis=1, how="all").dropna(how="any")

    # --- Retornos diarios (equivalente a pct_change) ---
    rets = returns_from_prices(prices).replace([np.inf, -np.inf], np.nan)

    # IMPORTANTÍSIMO: no uses how="any" demasiado pronto si te recorta.
    # Aquí ya tenemos panel completo por el dropna(how="any") en precios.
    rets = rets.dropna(how="any")

    # --- Inputs anualizados ---
    mu = rets.mean() * 252

    Sigma = risk_models.CovarianceShrinkage(
        rets, returns_data=True, frequency=252
    ).ledoit_wolf()
    Sigma = risk_models.fix_nonpositive_semidefinite(Sigma)

    # --- Solver disponible ---
    """
    solver, solver_options = pick_solver()
    """
    solvers = set(cp.installed_solvers())
    if "OSQP" in solvers:
        solver = "OSQP"
        solver_options = {"max_iter": 300000}
    elif "SCS" in solvers:
        solver = "SCS"
        solver_options = {"max_iters": 100000}
    else:
        raise RuntimeError(f"No tienes OSQP ni SCS. Solvers: {sorted(solvers)}")

    print(solver,solver_options)
   
    
    # --- Optimización ---
    # max_quadratic_utility maximiza: w^T mu - (delta/2) w^T Sigma w
    # tu objetivo: w^T mu - lam * w^T Sigma w  => delta = 2*lam
    delta = 2 * lam

    ef = EfficientFrontier(
        mu, Sigma,
        weight_bounds=(0, max_weight),   # clave para evitar 100% en un activo
        solver=solver,
        solver_options=solver_options
    )

    # Regularización opcional (ayuda a repartir pesos)
    if l2_gamma and l2_gamma > 0:
        ef.add_objective(objective_functions.L2_reg, gamma=l2_gamma)

    ef.max_quadratic_utility(risk_aversion=delta)

    w = pd.Series(ef.clean_weights(), index=mu.index).sort_values(ascending=False)

    # Top-k renormalizado (si quieres “10 valores”)
    w_top = w[w > 0].head(topk)
    if len(w_top) > 0:
        w_top = w_top / w_top.sum()

    return w, w_top, mu, Sigma, rets


# ===== USO =====
# prices = tu DF de precios (Adj Close), ya filtrado por NaNs/volumen como hiciste antes

w, w10, mu, Sigma, rets = markowitz_from_prices(
    prices,
    lam=10,          # prueba 10–30 si te concentra
    max_weight=0.20,   # 0.20–0.30 típico
    l2_gamma=0.3,     # si aún concentra, sube a 0.2–0.5
    topk=7
)

print("Pesos (top 15):")
display(w.head(15))

cuantos = 6
print(f"Top {cuantos} renormalizado:")
display(w10[:cuantos])
print(w10[:cuantos]*(100/sum(w10[:cuantos])))
```

```python
import numpy as np
from statsmodels.tsa.stattools import adfuller, kpss

def tests_estacionariedad(serie):
    x = np.asarray(serie, float)
    x = x[~np.isnan(x)]
    # --- ADF (H0: raíz unitaria / no estacionaria) ---
    adf_stat, adf_p, adf_lags, adf_n, adf_crit, adf_icbest = adfuller(x, autolag='AIC')
    print(f"ADF: stat={adf_stat:.3f}, p={adf_p:.4f}, lags={adf_lags}, n={adf_n}")
    #print("Crit values:", adf_crit)

    # --- KPSS (H0: estacionaria) ---
    # regression='c'  -> estacionaria alrededor de una constante (nivel)
    # regression='ct' -> estacionaria alrededor de una tendencia
    kpss_stat, kpss_p, kpss_lags, kpss_crit = kpss(x, regression='c', nlags='auto' )
    print(f"KPSS: stat={kpss_stat:.3f}, p={kpss_p:.4f}, lags={kpss_lags}")
    #print("Crit values:", kpss_crit)
    print("="*50)

tests_estacionariedad(prices["VWCE.DE"])
```

```python
import numpy as np
import pandas as pd

def portfolio_value_from_prices(prices: pd.DataFrame, w: pd.Series) -> pd.Series:
    """
    Construye una serie de 'valor de cartera' (empieza en 1.0) a partir de precios y pesos.
    """
    # Alinear tickers
    tickers = w.index.intersection(prices.columns)
    if len(tickers) == 0:
        raise ValueError("Ningún ticker de w está en prices.")

    w = w.loc[tickers].astype(float)
    w = w / w.sum()

    px = prices.loc[:, tickers].copy()
    px = px.replace([np.inf, -np.inf], np.nan).ffill().dropna(how="any")

    # Índice de cartera (base 1.0)
    port = (px / px.iloc[0]).mul(w, axis=1).sum(axis=1)
    port.name = "portfolio"
    return port

def annual_metrics(port: pd.Series) -> dict:
    """
    Devuelve CAGR y media anualizada (aritmética) usando retornos diarios.
    """
    port = port.dropna()
    rets = port.pct_change().dropna()

    # Años reales según fechas (más correcto que 252/len)
    years = (port.index[-1] - port.index[0]).days / 365.25
    cagr = (port.iloc[-1] / port.iloc[0])**(1/years) - 1 if years > 0 else np.nan

    ann_mean = rets.mean() * 252
    ann_vol  = rets.std()  * np.sqrt(252)

    return {"CAGR": cagr, "Media_anualizada": ann_mean, "Vol_anualizada": ann_vol}

def yearly_returns(port: pd.Series) -> pd.Series:
    """
    Retornos por año natural (cierre a cierre). Índice: año.
    """
    year_end = port.resample("Y").last()
    yr = year_end.pct_change().dropna()
    yr.index = yr.index.year
    yr.name = "retorno_anual"
    return yr

# =========================
# USO (últimos N años)
# =========================
N = 5  # últimos 3 años (cámbialo)
end = prices.index.max()
start = end - pd.DateOffset(years=N)

port = portfolio_value_from_prices(prices.loc[start:], w10[:4])

m = annual_metrics(port)
print(f"Periodo: {port.index.min().date()} -> {port.index.max().date()}")
print("CAGR:", round(m["CAGR"]*100, 2), "%")
print("Media anualizada (aritmética):", round(m["Media_anualizada"]*100, 2), "%")
print("Vol anualizada:", round(m["Vol_anualizada"]*100, 2), "%")

# Retornos por año natural y su media
yr = yearly_returns(port)
display(yr)

print("Media de retornos anuales (año natural):", round(yr.mean()*100, 2), "%")
```
