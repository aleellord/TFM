```python
import os
import pandas as pd

# =========================================================
# CONFIGURACIÓN
# =========================================================

DATA_DIR = "../Datos_csv"

files_semanal = [
    "df_semanal.csv",
    "df_semanal_1.csv",
    "df_semanal_2.csv",
    "df_semanal_3.csv",
    "df_semanal_4.csv",
]

files_diario = [
    "df_diario.csv",
    "df_diario_1.csv",
    "df_diario_2.csv",
    "df_diario_3.csv",
    "df_diario_4.csv",
]

metodos = [
    "LinearRegression",
    "Ridge",
    "RandomForestRegressor",
    "GradientBoostingRegressor",
]

# Orden deseado de métodos
orden_metodos = {
    "LinearRegression": 0,
    "Ridge": 1,
    "RandomForestRegressor": 2,
    "GradientBoostingRegressor": 3
}


# FUNCIÓN PARA DETECTAR TARGETS

def detectar_targets(df):
    cols = df.columns.tolist()
    targets = [col for col in cols if col.startswith("target_")]
    return targets



# FUNCIÓN GENERADORA

def generar_experimentos(files, output_file):
    rows = []

    for fname in files:
        path = os.path.join(DATA_DIR, fname)

        if not os.path.exists(path):
            print(f"No encontrado, se salta: {fname}")
            continue

        df = pd.read_csv(path, nrows=1)

        # targets disponibles
        targets = detectar_targets(df)

        if len(targets) == 0:
            print(f"Sin targets detectados en: {fname}")
            continue

        for metodo in metodos:
            for target in targets:
                rows.append({
                    "Archivo": fname,
                    "Método": metodo,
                    "Target": target,
                    "FechaIni": "",
                    "FechaFin": "",
                    "RMSE": "",
                    "RMSE_baseline": "",
                    "RMSE_zeros": "",
                    "MAE": "",
                    "MAE_baseline": "",
                    "MAE_zeros": "",
                    "Estado": "PENDIENTE",
                    "Error": ""
                })

    # Crear DataFrame
    experimentos = pd.DataFrame(rows)

    if experimentos.empty:
        print(f"No se ha podido crear {output_file} porque no hay experimentos.")
        return

    # ORDENAR POR ARCHIVO, MÉTODO Y TARGET

    # Añadir columna auxiliar para ordenar
    experimentos["orden_metodo"] = experimentos["Método"].map(orden_metodos)

    # Ordenar: primero archivo, luego método, luego target
    experimentos = experimentos.sort_values(
        by=["Archivo", "orden_metodo", "Target"]
    ).drop(columns="orden_metodo")

    # Resetear índice
    experimentos = experimentos.reset_index(drop=True)

    # GUARDAR
    experimentos.to_csv(output_file, index=False, encoding="utf-8-sig")

    print(f"Creado {output_file} con {len(experimentos)} filas.")
    print(experimentos.head(10))
    print("-" * 60)


# GENERAR EXPERIMENTOS

generar_experimentos(files_semanal, "Experimentos_semanales.csv")
generar_experimentos(files_diario, "Experimentos_diarios.csv")
```
