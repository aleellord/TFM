# Extracción de datos

Descarga precios ajustados y volumen histórico de 63 ETFs desde Yahoo Finance para el periodo 2021–2026. Los datos se guardan como CSV en la carpeta `Datos_csv/` y sirven de base para todos los notebooks posteriores.

```python
import yfinance as yf
import pandas as pd
import numpy as np

TICKERS = [
    # US Core
    "SPY","IVV","VOO","QQQ","VTI","ITOT","DIA","IWM","IJR","MDY",
    # Styles / Factors
    "VUG","IWF","VTV","IWD","SCHD","VIG","DVY","MTUM","QUAL","USMV","VLUE",
    # Sectors
    "XLK","XLF","XLV","XLY","XLP","XLE","XLI","XLU","XLB","XLRE",
    # International
    "VEA","IEFA","VWO","IEMG","EFA","EEM","EWJ","EWG","EWU","EWQ",
    "INDA","EWZ","FXI","MCHI","EWT","EWY",
    # Bonds
    "BND","AGG","IEF","TLT","SHY","LQD","HYG","JNK","TIP","BIL",
    # Commodities / REITs
    "GLD","IAU","SLV","USO","DBC","VNQ",
]


```

```python
# Descargar datos históricos
data = yf.download(TICKERS, start="2021-01-01", auto_adjust=False, progress=False)
```

```python
adj_close = data["Adj Close"]
# Eliminar el nombre del nivel 
adj_close.columns.name = None
volume = data["Volume"]
volume.columns.name = None
```

```python
import os

# Carpeta destino (Datos_csv, fuera de 1.Preprocesado)
out_dir = "../Datos_csv"
os.makedirs(out_dir, exist_ok=True)

# Rutas de salida
path_adj_close = os.path.join(out_dir, "adj_close.csv")
path_volume = os.path.join(out_dir, "volume.csv")

# Guardar CSVs
adj_close.to_csv(path_adj_close, index=True)
volume.to_csv(path_volume, index=True)

# Confirmación
print("Guardado en:")
print(" -", path_adj_close, adj_close.shape)
print(" -", path_volume, volume.shape)

```
